##   Visualization

```python
monthly_sales = df.groupby('Month')['SALES'].sum()
plt.figure(figsize=(8,5))
monthly_sales.sort_index().plot(kind='bar')
plt.title("Monthly Salse")
plt.xlabel("month")
plt.ylabel("salse")
plt.show()


```

```python
country_sales = df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
country_sales.head(10).plot(kind='bar')
plt.title("Top 10 Countries by Sales")
plt.xlabel("Country")
plt.ylabel("Sales")
plt.show()
```

```python
product_sales = df.groupby('PRODUCTLINE')['SALES'].sum().sort_values()

plt.figure(figsize=(8,5))
product_sales.plot(kind='barh')
plt.title("Sales by Product Line")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.show()
```

```python
cancel_rate = df.groupby('PRODUCTLINE')['STATUS']  .apply(lambda x: (x != 'Shipped').mean()).sort_values()

plt.figure(figsize=(8,5))
cancel_rate.plot(kind='barh')
plt.title("Cancellation Rate by Product")
plt.xlabel("Rate")
plt.ylabel("Product")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['SALES'].hist()
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
